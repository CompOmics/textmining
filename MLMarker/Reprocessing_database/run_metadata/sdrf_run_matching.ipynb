# Function for fetching SDRF

```python
SDRF_CACHE_DIR = r"C:\Users\sander\OneDrive\Bureaublad\Projects\agentic-metadata\run_level_data\sdrf_cache"
os.makedirs(SDRF_CACHE_DIR, exist_ok=True)


def fetch_sdrf(pxd_id: str, timeout: int = 240, use_cache: bool = True) -> pd.DataFrame | None:
    """Fetch SDRF table for a PXD accession from PRIDE.

    If use_cache is True, saves to / loads from local cache on first fetch.
    Returns a DataFrame if SDRF exists, None otherwise.
    """
    # Check local cache first
    cache_path = os.path.join(SDRF_CACHE_DIR, f"{pxd_id}.sdrf.tsv")
    if use_cache and os.path.exists(cache_path):
        return pd.read_csv(cache_path, sep="\t")

    # Find SDRF file in PRIDE file list
    url = f"https://www.ebi.ac.uk/pride/ws/archive/v3/projects/{pxd_id}/files"
    page = 0
    sdrf_file = None

    while page < 50:
        try:
            resp = requests.get(url, params={"pageSize": 100, "page": page}, timeout=30)
            resp.raise_for_status()
            files = resp.json()
            if not files:
                break
            for f in files:
                name = f.get("fileName", "")
                if name.lower() == "sdrf.tsv" or name.lower().endswith(".sdrf.tsv"):
                    sdrf_file = f
                    break
            if sdrf_file or len(files) < 100:
                break
            page += 1
        except Exception as e:
            print(f"  {pxd_id}: file list error: {e}")
            break

    if not sdrf_file:
        return None

    download_url = None
    for loc in sdrf_file.get("publicFileLocations", []):
        u = loc.get("value", "")
        if u.startswith("ftp://"):
            download_url = u.replace(
                "ftp://ftp.pride.ebi.ac.uk/pride/data/archive/",
                "https://ftp.pride.ebi.ac.uk/pride/data/archive/"
            )
            break
        elif u.startswith("http"):
            download_url = u
            break

    if not download_url:
        return None

    try:
        resp = requests.get(download_url, timeout=timeout)
        resp.raise_for_status()
        df = pd.read_csv(StringIO(resp.text), sep="\t")

        # Save to cache
        if use_cache:
            df.to_csv(cache_path, sep="\t", index=False)

        return df
    except Exception as e:
        print(f"  {pxd_id}: SDRF download failed: {e}")
        return None
```

# SDRF -> Run-level metadata extraction

```python
import re, os

# --- SDRF column -> metadata field mapping ---
SDRF_COL_MAP = {
    "characteristics[organism]":       "organism",
    "characteristics[organism part]":  "tissue",
    "characteristics[disease]":        "disease",
    "characteristics[cell line]":      "cell_line",
    "comment[instrument]":             "instrument",
    "comment[dissociation method]":    "fragmentation",
    "comment[collision energy]":       "collision_energy",
    "comment[cleavage agent details]": "enzymes",
    "comment[modification parameters]":"modifications",
    "comment[label]":                  "labeling",
    "characteristics[enrichment process]": "enrichment",
}

# Load known organisms for organism/tissue swap detection
_ORGANISMS_PATH = r"C:\Users\sander\OneDrive\Bureaublad\Projects\agentic-metadata\ontologies\organisms.txt"
KNOWN_ORGANISMS = set()
with open(_ORGANISMS_PATH, encoding="utf-8") as _f:
    for _line in _f:
        t = _line.strip()
        if t and not t.startswith("#"):
            KNOWN_ORGANISMS.add(t.lower())


def _parse_sdrf_value(val) -> str | None:
    """Extract the NT= (name) portion from an SDRF ontology-annotated value."""
    if pd.isna(val):
        return None
    val = str(val).strip()
    if not val or val.lower() in ("not available", "not applicable"):
        return None
    m = re.search(r"NT=([^;]+)", val)
    if m:
        return m.group(1).strip()
    return val


def _stem_filename(name: str) -> str:
    """Strip common raw-file extensions to get a comparable run stem."""
    return re.sub(r"\.(raw|mzML|d|wiff)$", "", name, flags=re.IGNORECASE)


def _normalize_sdrf_columns(sdrf: pd.DataFrame) -> dict:
    """Build a lowercase column-name -> original column-name lookup."""
    return {c.lower(): c for c in sdrf.columns}


def _find_duplicate_columns(sdrf: pd.DataFrame, sdrf_key: str) -> list[str]:
    """Find the base column + any pandas-suffixed duplicates (.1, .2, ...)."""
    return [c for c in sdrf.columns if c.lower() == sdrf_key or re.match(rf"^{re.escape(sdrf_key)}\.\d+$", c.lower())]


def _fix_organism_tissue_swap(df: pd.DataFrame) -> pd.DataFrame:
    """Fix cases where SDRF has organism and tissue columns swapped."""
    if "organism" not in df.columns or "tissue" not in df.columns:
        return df
    df = df.copy()
    tissue_is_organism = df["tissue"].apply(
        lambda v: v.lower() in KNOWN_ORGANISMS if pd.notna(v) else False
    )
    organism_is_missing = df["organism"].isna()
    swap_mask = tissue_is_organism & organism_is_missing
    if swap_mask.any():
        df.loc[swap_mask, "organism"] = df.loc[swap_mask, "tissue"]
        df.loc[swap_mask, "tissue"] = None
        print(f"  Fixed organism/tissue swap for {swap_mask.sum()} rows")
    return df


def _fill_enrichment_fallback(df: pd.DataFrame, sdrf: pd.DataFrame, col_lookup: dict) -> pd.DataFrame:
    """Fill enrichment from factor value[enrichment process] if characteristics was empty."""
    fv_col = col_lookup.get("factor value[enrichment process]")
    if fv_col is None or "enrichment" not in df.columns:
        return df
    df = df.copy()
    for idx in df.index:
        if pd.isna(df.at[idx, "enrichment"]) and idx < len(sdrf):
            fv = sdrf.iloc[idx][fv_col]
            if pd.notna(fv) and str(fv).strip().lower() not in ("not available", "not applicable", ""):
                df.at[idx, "enrichment"] = str(fv).strip()
    return df


def _normalize_enrichment(df: pd.DataFrame) -> pd.DataFrame:
    """Normalize enrichment: if value contains 'phospho', set to 'phosphorylation'."""
    if "enrichment" not in df.columns:
        return df
    df = df.copy()
    df["enrichment"] = df["enrichment"].apply(
        lambda v: "phosphorylation" if pd.notna(v) and "phospho" in v.lower() else v
    )
    return df


def _detect_fractionation(sdrf: pd.DataFrame, col_lookup: dict) -> bool:
    """Detect if the experiment uses fractionation from SDRF columns.

    Returns True if:
      - comment[fraction identifier] has more than 1 unique value, OR
      - comment[fractionation method] has a real value (not 'not available')
    """
    frac_id_col = col_lookup.get("comment[fraction identifier]")
    if frac_id_col is not None:
        unique_ids = sdrf[frac_id_col].dropna().unique()
        if len(unique_ids) > 1:
            return True

    frac_method_col = col_lookup.get("comment[fractionation method]")
    if frac_method_col is not None:
        methods = sdrf[frac_method_col].dropna().unique()
        real_methods = [m for m in methods if str(m).strip().lower() not in ("not available", "not applicable", "")]
        if real_methods:
            return True

    return False


def sdrf_to_run_metadata(sdrf: pd.DataFrame, pxd_id: str) -> pd.DataFrame:
    """Convert an SDRF table into per-run metadata rows."""
    col_lookup = _normalize_sdrf_columns(sdrf)
    data_file_col = col_lookup.get("comment[data file]")
    if data_file_col is None:
        raise ValueError("SDRF has no 'comment[data file]' column")

    rows = []
    for _, srow in sdrf.iterrows():
        run_name = _stem_filename(str(srow[data_file_col]))
        row = {"pxd": pxd_id, "run": run_name}
        for sdrf_key, meta_field in SDRF_COL_MAP.items():
            orig_col = col_lookup.get(sdrf_key)
            if orig_col is None:
                row[meta_field] = None
                continue
            dup_cols = _find_duplicate_columns(sdrf, sdrf_key)
            if not dup_cols:
                dup_cols = [orig_col]
            parsed_vals = []
            for dc in dup_cols:
                v = _parse_sdrf_value(srow[dc])
                if v and v not in parsed_vals:
                    parsed_vals.append(v)
            row[meta_field] = "; ".join(parsed_vals) if parsed_vals else None
        # Fraction identifier (raw value, not parsed through NT=)
        frac_col = col_lookup.get("comment[fraction identifier]")
        if frac_col is not None:
            frac_val = srow.get(frac_col)
            row["fraction_identifier"] = int(frac_val) if pd.notna(frac_val) else None
        else:
            row["fraction_identifier"] = None

        rows.append(row)

    df = pd.DataFrame(rows)
    df = _fix_organism_tissue_swap(df)
    df = _fill_enrichment_fallback(df, sdrf, col_lookup)
    df = _normalize_enrichment(df)

    # Fractionation: project-level boolean
    is_fractionated = _detect_fractionation(sdrf, col_lookup)
    df["fractionation"] = True if is_fractionated else None

    return df


# --- Test with PXD008012 ---
run_meta = sdrf_to_run_metadata(sdrf, "PXD008012")
print(f"Run-level metadata: {len(run_meta)} rows, {len(run_meta.columns)} columns")
display(run_meta.head(10))
```

# Normalize SDRF metadata through the normalization pipeline
Same SapBERT + post-normalization map used for the LLM extraction output.

```python
import sys, importlib

PROJECT_ROOT = r"C:\Users\sander\OneDrive\Bureaublad\Projects\agentic-metadata"
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import normalization.normalize_extraction
importlib.reload(normalization.normalize_extraction)

from normalization.config import NormalizationConfig
from normalization.normalizer import TermNormalizer
from normalization.normalize_extraction import (
    CellLineExactMatcher, EVIDENCE_ARRAY_FIELDS,
    _load_post_normalization_map, _apply_post_normalization,
)

norm_config = NormalizationConfig.from_dict({
    "ontology_dir": os.path.join(PROJECT_ROOT, "ontologies"),
    "cache_dir": os.path.join(PROJECT_ROOT, "ontology_cache"),
    "use_gpu": True,
})
normalizer = TermNormalizer(norm_config)
normalizer.load_all_ontologies()

cell_line_path = norm_config.get_ontology_path("cell_lines")
cell_line_matcher = CellLineExactMatcher(cell_line_path) if cell_line_path and cell_line_path.exists() else None

post_map = _load_post_normalization_map(norm_config.ontology_dir)

print(f"Loaded {len(normalizer.indices)} ontologies")
print(f"Post-normalization map: {post_map}")
if cell_line_matcher:
    print(f"Cell line matcher: {len(cell_line_matcher.canonical)} entries")
```

```python
# Fields from SDRF_COL_MAP that have a corresponding SapBERT ontology
NORMALIZABLE_FIELDS = {
    field: entity_type
    for field, entity_type in EVIDENCE_ARRAY_FIELDS.items()
    if field in SDRF_COL_MAP.values()
}

# Fields where unmatched values should be dropped (set to None).
# These must match controlled vocabulary; unknown values are not useful.
STRICT_FIELDS = {"organism", "tissue", "disease", "instrument", "fragmentation", "enzymes"}

def normalize_run_metadata(df: pd.DataFrame, normalizer: TermNormalizer,
                           cell_line_matcher: CellLineExactMatcher,
                           post_map: dict) -> pd.DataFrame:
    """Normalize SDRF-derived run metadata through the same pipeline as LLM extractions.

    For multi-value fields (e.g. "Trypsin; Lys-C"), each value is normalized individually.
    Duplicate values after normalization are removed.
    For strict fields, unmatched values are dropped.
    """
    df = df.copy()

    for field, entity_type in NORMALIZABLE_FIELDS.items():
        if field not in df.columns:
            continue
        strict = field in STRICT_FIELDS
        normalized = []
        for val in df[field]:
            if pd.isna(val):
                normalized.append(None)
                continue
            parts = [v.strip() for v in str(val).split(";")]
            norm_parts = []
            seen = set()
            for part in parts:
                if not part:
                    continue
                result = normalizer.normalize(part, entity_type=entity_type)
                if result.is_normalized:
                    item = {"ontology_name": result.ontology_name}
                    _apply_post_normalization(item, entity_type, post_map)
                    out = item["ontology_name"]
                else:
                    if strict:
                        continue
                    out = part
                if out.lower() not in seen:
                    norm_parts.append(out)
                    seen.add(out.lower())
            normalized.append("; ".join(norm_parts) if norm_parts else None)
        df[field] = normalized

    # Cell line exact matching
    if cell_line_matcher and "cell_line" in df.columns:
        df["cell_line"] = df["cell_line"].apply(
            lambda v: cell_line_matcher.match(v) or v if pd.notna(v) else None
        )

    return df


run_meta_normalized = normalize_run_metadata(run_meta, normalizer, cell_line_matcher, post_map)

# Show disease normalization changes
comparison = pd.DataFrame({
    "before": run_meta["disease"],
    "after": run_meta_normalized["disease"],
})
changed = comparison[comparison["before"] != comparison["after"]].drop_duplicates()
print("Disease normalization changes:")
display(changed if len(changed) else "No changes")

display(run_meta_normalized.head(5))
```

# Multi-project SDRF extraction
Fetch SDRFs for a batch of projects, parse to run-level metadata, normalize, and combine.

```python
import time, json as _json

# Find projects that have SDRFs from local metadata
meta_dir = r"C:\Users\sander\OneDrive\Bureaublad\Projects\agentic-metadata\results\manuscripts\metadata"
sdrf_pxds = []
for fname in sorted(os.listdir(meta_dir)):
    with open(os.path.join(meta_dir, fname)) as fh:
        data = _json.load(fh)
    if data.get("has_sdrf"):
        sdrf_pxds.append(fname.replace(".meta.json", ""))

print(f"Projects with SDRF: {len(sdrf_pxds)}")
print(f"Loading SDRFs (cached or from PRIDE)...")

sdrf_results = {}
for i, pxd_id in enumerate(sdrf_pxds):
    cache_path = os.path.join(SDRF_CACHE_DIR, f"{pxd_id}.sdrf.tsv")
    from_cache = os.path.exists(cache_path)
    sdrf_df = fetch_sdrf(pxd_id)
    if sdrf_df is not None:
        sdrf_results[pxd_id] = sdrf_df
        src = "cache" if from_cache else "PRIDE"
        print(f"  [{i+1}/{len(sdrf_pxds)}] {pxd_id}: {len(sdrf_df)} rows ({src})")
    else:
        print(f"  [{i+1}/{len(sdrf_pxds)}] {pxd_id}: fetch failed")
    if not from_cache:
        time.sleep(2)

print(f"\nSuccessfully loaded: {len(sdrf_results)}/{len(sdrf_pxds)}")
```

```python
# Parse all SDRFs to run-level metadata
all_run_meta = []
for pxd_id, sdrf_df in sdrf_results.items():
    try:
        rm = sdrf_to_run_metadata(sdrf_df, pxd_id)
        all_run_meta.append(rm)
        print(f"  {pxd_id}: {len(rm)} runs extracted")
    except Exception as e:
        print(f"  {pxd_id}: FAILED - {e}")

run_meta_all = pd.concat(all_run_meta, ignore_index=True) if all_run_meta else pd.DataFrame()
print(f"\nTotal run-level rows: {len(run_meta_all)}")
print(f"Projects with data: {run_meta_all['pxd'].nunique()}")
display(run_meta_all.head(10))
```
