# Analyze disagreements

## Organisms

```python
# Rows where organism has disagreement (evidence contains '-')                                                                                                                  
organism_disagree = df[df["organism_evidence"].str.contains("-", na=False)]                                                                                                     
print(f"Organism disagreements: {len(organism_disagree)}\n")                                                                                                                                                                                                                                                                                                      
print(organism_disagree[["pxd", "run", "organism", "organism_evidence"]].to_string()) 
```

### Tissues

```python
# Tissue disagreements specifically between name and agent                                                                                                                      
tissue_name_vs_agent = df[df["tissue_evidence"] == "name; -agent"]
print(f"Tissue 'name; -agent' disagreements: {len(tissue_name_vs_agent)}\n")                                                                                                                                                                                                                                                                                      
print(tissue_name_vs_agent[["pxd", "run", "tissue", "tissue_evidence"]].head(400).to_string())                                                                                   

# What tissues did name extract vs what agent had?
print(f"\nTissue values (from name, winning):")
print(tissue_name_vs_agent["tissue"].value_counts().to_string())
```
