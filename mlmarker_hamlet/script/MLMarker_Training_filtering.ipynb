# MLMarker Training Datasets Filtering

This notebook loads the 183 PRIDE projects (PXDs) used to train the MLMarker model (from `SupplementaryTableS1.tsv`).



```python
import os 
import pandas as pd
```

```python
# PATH_MLMARKER = r"C:\Users\jung.arnaud\Project\HAMLET_MLMarker\mlmarker_pipeline\results_0909\run_meta_mlmarker_all_sample_nsaf_matrix.tsv"
PATH_MLMARKER = r"C:\Users\jung.arnaud\Project\HAMLET_MLMarker\mlmarker_pipeline\results_1109\run_meta_mlmarker_sample_nsaf_matrix.tsv"
PATH_TRAINING = r"C:\Users\jung.arnaud\Git_Folder\textmining\MLMarker\Training_PXDs\SupplementaryTableS1.tsv"
PATH_OUTPUT   = r"C:\Users\jung.arnaud\Project\HAMLET_MLMarker\mlmarker_pipeline\results_1109"
```

```python
df_mlmarker = pd.read_csv(PATH_MLMARKER, sep='\t', header=0)
df_mlmarker
```

```python
with open(PATH_TRAINING, "r", encoding="utf-8") as f:
    training_pxds = [
        token for token in f.read().split() 
        if token.startswith("PXD") and token != "PXD_accession"
    ]
print(f"Total training PXDs loaded: {len(training_pxds)}")
print(f"First 10 PXDs: {training_pxds[:10]}")

df_training = pd.DataFrame(training_pxds, columns=["pxd"])
df_training
```

```python
# Example: filter an existing dataset (e.g., df_metadata) to exclude MLMarker training projects
df_unseen = df_mlmarker[~df_mlmarker["pxd"].isin(training_pxds)]
print(f"Kept {len(df_unseen)} unseen samples out of {len(df_mlmarker)}")
```

```python
unseen_pxd = df_unseen['pxd'].unique()
unseen_pxd
```
